A. Charger les fichiers

```python
import pandas as pd

traffic = pd.read_csv("traffic_cleaned.csv")
population = pd.read_csv("../data_clean/population_clean.csv")
roads = pd.read_csv("../data_clean/roads_clean.csv")
transport = pd.read_csv("../data_clean/transport_clean.csv")
landuse = pd.read_csv("../data_clean/landuse_clean.csv")
```

B. Créer les dimensions

Dimension Temps

```python
dim_day = traffic[["day"]].drop_duplicates()
dim_day["day_id"] = dim_day.index
```

Dimension Commune

```python
dim_commune = traffic[["COMMUNE"]].drop_duplicates()
dim_commune["commune_id"] = dim_commune.index
```

Dimension Route

```python
dim_route = traffic[[
    "Route",
    "Latitude Debut",
    "Longitude Debut",
    "Latitude Fin",
    "Longitude Fin"
]].drop_duplicates()

dim_route["route_id"] = dim_route.index
```
