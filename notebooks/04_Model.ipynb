## 04_Model.ipynb

A. Charger les fichiers

```python
import pandas as pd

traffic = pd.read_csv("traffic_cleaned.csv")
population = pd.read_csv("../data_clean/population_clean.csv")
roads = pd.read_csv("../data_clean/roads_clean.csv")
transport = pd.read_csv("../data_clean/transport_clean.csv")
landuse = pd.read_csv("../data_clean/landuse_clean.csv")
```

B. Créer les dimensions

Dimension Temps

```python
dim_day = traffic[["day"]].drop_duplicates()
dim_day["day_id"] = dim_day.index
```

Dimension Commune

```python
dim_commune = traffic[["COMMUNE"]].drop_duplicates()
dim_commune["commune_id"] = dim_commune.index
```

Dimension Route

```python
dim_route = traffic[[
    "Route",
    "Latitude Debut",
    "Longitude Debut",
    "Latitude Fin",
    "Longitude Fin"
]].drop_duplicates()

dim_route["route_id"] = dim_route.index
```

🟢 1️⃣ CRÉER LA COLONNE ROUTE

```python
traffic["Route"] = (
    traffic["Latitude Debut"].round(3).astype(str)
    + "," +
    traffic["Longitude Debut"].round(3).astype(str)
    + " → " +
    traffic["Latitude Fin"].round(3).astype(str)
    + "," +
    traffic["Longitude Fin"].round(3).astype(str)
)
```

```python
dim_route = traffic[[
    "Route",
    "Latitude Debut",
    "Longitude Debut",
    "Latitude Fin",
    "Longitude Fin"
]].drop_duplicates()

dim_route["route_id"] = dim_route.index
```

```python
##VÉRIFIER
traffic.columns
```

C. Créer la table FACT (le cœur)

```python
fact_traffic = traffic.merge(dim_day, on="day")
fact_traffic = fact_traffic.merge(dim_commune, on="COMMUNE")
fact_traffic = fact_traffic.merge(dim_route, on="Route")

fact_traffic = fact_traffic[[
    "day_id",
    "commune_id",
    "route_id",
    "tti_mean",
    "Distance en KM"
]]
```

D. Sauvegarder

```python
dim_day.to_csv("../data_clean/dim_day.csv", index=False)
dim_commune.to_csv("../data_clean/dim_commune.csv", index=False)
dim_route.to_csv("../data_clean/dim_route.csv", index=False)
fact_traffic.to_csv("../data_clean/fact_traffic.csv", index=False)
```
