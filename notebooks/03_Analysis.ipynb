4️⃣ Communes les plus congestionnées Avec La presentation Graphique

```python
traffic.groupby("COMMUNE")["tti_mean"].mean().sort_values(ascending=False)
```

```python
import matplotlib.pyplot as plt

traffic.groupby("COMMUNE")["tti_mean"].mean().sort_values().plot(kind="barh")
plt.title("Congestion moyenne par commune")
plt.show()
```

5️⃣ Congestion par jour Avec La presentation Graphique

```python
traffic.groupby("day")["tti_mean"].mean()
```

```python
traffic.groupby("day")["tti_mean"].mean().plot(kind="line", marker="o")
plt.title("Congestion par jour")
plt.show()
```

6️⃣ Routes les plus congestionnées Avec la presentation Graphique

```python
traffic.sort_values("tti_mean", ascending=False).head(10)
```

```python
traffic.sort_values("tti_mean", ascending=False).head(10).plot(
    x="index",
    y="tti_mean",
    kind="bar"
)
plt.title("Top routes congestionnées")
plt.show()
```
